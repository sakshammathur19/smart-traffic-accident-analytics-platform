```python
import boto3

AWS_ACCESS_KEY = "YOUR_AWS_ACCESS_KEY"
AWS_SECRET_KEY = "YOUR_AWS_SECRET_KEY"


TOPIC_ARN = "arn:aws:sns:eu-north-1:465861111051:smart-traffic-alerts"

sns = boto3.client(
    "sns",
    region_name="eu-north-1",
    aws_access_key_id=AWS_ACCESS_KEY,
    aws_secret_access_key=AWS_SECRET_KEY
)
```

```python
alerts = spark.sql("""
SELECT *
FROM gold_emergency_alerts
""")

display(alerts)
```

```python
rows = alerts.collect()

print(len(rows))
```

```python
for row in rows:

    message = f"""
SMART TRAFFIC ALERT

Road : {row.road_name}

Priority : {row.priority}

Vehicle Count : {row.vehicle_count}

Average Speed : {row.avg_speed}

Time : {row.timestamp}
"""

    sns.publish(
        TopicArn=TOPIC_ARN,
        Subject="Critical Traffic Alert",
        Message=message
    )

print("Alerts Sent Successfully")
```
